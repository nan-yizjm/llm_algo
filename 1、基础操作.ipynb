## 1. 张量维度变换 (Tensor Reshaping)

### 1.1 原生操作

```python
x = torch.arange(24).reshape(2, 3, 4)
print('原始 shape：', x.shape)

# 1. view() 方法
# view() 方法用于改变张量的形状，但不改变其数据。它返回一个新的张量，具有相同的数据，但形状不同。使用 view() 方法时，需要确保新形状的元素总数与原始张量的元素总数相同。
print('view() 方法改变形状：', x.view(6, 4).shape)
print('内存连续性：', x.view(6, 4).is_contiguous())

# 2. reshape() 方法
# reshape() 方法与 view() 方法类似，也用于改变张量的形状。它返回一个新的张量，具有相同的数据，但形状不同。与 view() 方法不同的是，reshape() 方法可以处理非连续的张量。
print('reshape() 方法改变形状：', x.reshape(6, 4).shape)
print('内存连续性：', x.reshape(6, 4).is_contiguous())

# 3. permute() 方法
# permute() 方法用于改变张量的维度顺序。它返回一个新的张量，具有相同的数据，但维度顺序不同。使用 permute() 方法时，需要指定新的维度顺序。
print('permute() 方法改变维度顺序：', x.permute(1, 0, 2).shape)
print('内存连续性：', x.permute(1, 0, 2).is_contiguous())

# 4. transpose() 方法
# transpose() 方法用于交换张量的两个维度。它返回一个新的张量，具有相同的数据，但指定的两个维度被交换。使用 transpose() 方法时，需要指定要交换的两个维度。
print('transpose() 方法交换维度：', x.transpose(0, 1).shape)
print('内存连续性：', x.transpose(0, 1).is_contiguous())
```

### 1.2 einops实现
einops是一个用于张量操作的Python库，它提供了简洁且强大的API来进行张量的重排、重塑和组合。通过einops，我们可以轻松地实现复杂的张量操作，而无需编写冗长的代码。
实现张量维度变换的主要函数是`rearrange`，它允许我们以一种声明性的方式指定输入张量的维度如何重新排列为输出张量的维度。各个参数含义如下：
- `input_tensor`：输入张量，通常是一个多维数组或张量

- `**axes_lengths`：可选的关键字参数，用于指定某些维度的长度。如果在模式中使用了字母表示维度，但没有提供对应的长度，einops会尝试从输入张量中推断这些维度的长度。如果无法推断，或者需要显式指定某些维度的长度，可以通过`**axes_lengths`参数传递。例如，`rearrange(input_tensor, 'b c h w -> b (c h w)', h=32, w=32)`将显式指定高度和宽度为32。

## 3. 前向传播与反向传播 (Forward & Backward)

**前向传播 (Forward Pass)：** 数据从输入层流向输出层，经过一系列的线性变换和非线性激活函数。在这个过程中，我们需要保存中间结果（如激活值、mask 等），供反向传播使用。

**反向传播 (Backward Pass)：** 梯度从输出层反向流向输入层，利用链式法则逐层计算每个参数的梯度。这是深度学习训练的核心机制。

```python
class LinearReLUFunction(torch.autograd.Function):
    """
    实现一个包含 Linear + ReLU 的算子，并推导其反向传播的梯度。
    公式: y = relu(x @ W^T + b)
    """
    
    @staticmethod
    def forward(ctx, x, weight, bias):
        # ==========================================
        # TODO 3.1: 实现前向传播
        # 1. 使用 F.linear 计算线性变换
        # 2. 使用 F.relu 计算激活
        # 3. 计算并保存 mask，用于反向传播时判断哪些位置需要传递梯度
        # 4. 保存必要的张量供反向传播使用
        # ==========================================
        # z = ???
        # y = ???
        # mask = ???
        # ctx.save_for_backward(???)
        z=F.linear(x,weight,bias)
        y=F.relu(z)
        mask=(z>0).float()
        ctx.save_for_backward(x,weight,mask)
        return y

    @staticmethod
    def backward(ctx, grad_output):
        """
        接收从上一层传回来的梯度 (grad_output)，形状同 y。
        返回对当前层三个输入 (x, weight, bias) 的梯度。
        """
        x, weight, mask = ctx.saved_tensors
        
        # ==========================================
        # TODO 3.2: 反传过 ReLU
        # 提示: ReLU 的导数在正值处为 1，负值处为 0
        # ==========================================
        # grad_z = ???
        grad_z=grad_output*mask
        
        # ==========================================
        # TODO 3.3: 反传过 Linear
        # 提示: 利用矩阵求导的链式法则，分别计算对 x, weight, bias 的梯度
        # 注意矩阵维度的匹配和转置操作
        # ==========================================
        # grad_x = ???
        # grad_weight = ???
        # grad_bias = ???
        grad_x=grad_z@weight
        grad_weight=grad_z.T@x
        grad_bias=grad_z.sum(dim=0)
        
        return grad_x, grad_weight, grad_bias
```
